## **ESERCIZIO 6 – PERIMETRO E AREA DI UN RETTANGOLO (v2)**
Scrivi un programma che, dopo aver richiesto la misura dell'altezza di un rettangolo, ne determini perimetro e area, sapendo che la base misura il doppio dell'altezza


```python
altezza = float(input("Inserisci l'altezza del rettangolo: "))
base = 2 * altezza
area = base * altezza
perimetro = 2 * (base + altezza)
print("Base:", base)
print("Altezza:", altezza)
print("Area:", area)
print("Perimetro:", perimetro)
```

## **ESERCIZIO 7 – PERIMETRO E AREA DI UN TRAPEZIO**
Determina l’area di un trapezio leggendo in input il valore delle basi e l’altezza.


```python
dati = input().split()
base_maggiore = float(dati[0])
base_minore = float(dati[1])
altezza = float(dati[2])
area = (base_maggiore + base_minore) * altezza / 2
print(area)
```

## **ESERCIZIO 11 – IPOTENUSA DI UN TRIANGOLO RETTANGOLO**
In un triangolo rettangolo, calcola la misura dell’ipotenusa, date le misure dei due cateti


Aiuto:


*   occorre importare la libreria math, scrivendo nella prima riga: `import math`
*   per calcolare la radice quadrata, occorre richiamare la funzione: `math.sqrt()` e all'interno della parentesi scrivere l'espressione matematica che andrebbe sotto la radice



```python
import math
riga = input()
valori = riga.split()
cateto1 = float(valori[0])
cateto2 = float(valori[1])
ipotenusa = math.sqrt(cateto1 * cateto1 + cateto2 * cateto2)
print(ipotenusa)
```

## **ESERCIZIO 13 – CONVERSIONE IN MINUTI**
Scrivi un programma che, data in input la misura di un intervallo di tempo espressa in ore e minuti, converta tutto in minuti.
Per esempio, 1 ora e 10 minuti sono 70 minuti.


```python
linea = input()
parti = linea.split()
if len(parti) == 2:
    ore = int(parti[0])
    minuti = int(parti[1])
else:
    ore = int(parti[0])
    minuti = int(input())
totale_minuti = ore * 60 + minuti
print(totale_minuti)
```

## **ESERCIZIO 19 – CONVERSIONE IN FAHRENHEIT**
Scrivi un programma che converta una temperatura data in gradi Celsius in gradi Fahrenheit. (Calcola la temperatura equivalente in gradi Fahrenheit utilizzando la formula F=9/5×C+32, dove C è la temperatura in gradi Celsius).


```python
valore_input = input()
temperatura_celsius = float(valore_input)
coefficiente = 9.0 / 5.0
parte_intermedia = coefficiente * temperatura_celsius
temperatura_fahrenheit = parte_intermedia + 32.0
print(temperatura_fahrenheit)
```
