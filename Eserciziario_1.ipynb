```python
linea = input()
parti = linea.split()
if len(parti) == 2:
    ore = int(parti[0])
    minuti = int(parti[1])
else:
    ore = int(parti[0])
    minuti = int(input())
totale_minuti = ore * 60 + minuti
print(totale_minuti)
```

# **CICLO WHILE**

## **ESERCIZIO 1 - SOMMA FINO A ZERO**
Leggi numeri finché l'utente inserisce 0 e stampa la somma di tutti i numeri inseriti.

```python
somma = 0
n = int(input())
while n != 0:
    somma += n
    n = int(input())
print(somma)
```

## **ESERCIZIO 2 - STAMPA I MULTIPLI DI 3**
Stampa i multipli di 3 minori di 30.

```python
numero = 3
while numero < 30:
    print(numero)
    numero += 3
```

# **CICLO FOR**

## **ESERCIZIO 1 - STAMPA I NUMERI DA 1 A 10**

## **ESERCIZIO 2 - SOMMA DEI PRIMI 10 NUMERI**

```python
somma = 0
for i in range(1, 11):
    somma += i
print(somma)
```
